### ✅ Task 18 — Sorting
Given:
```python
arr = np.array([3, 1, 9, 5, 2, 8, 4, 7, 6])
```
1. Sort the array in **ascending order**.
2. Sort in **descending order**.
3. Return the **indices that would sort** the array (`np.argsort`).
4. Find the index of the **maximum** and **minimum** values.

````python
# ✏️ Write your answer here
arr = np.array([3, 1, 9, 5, 2, 8, 4, 7, 6])
arr1=arr.sort()#Sorts the array in-place (modifies the original array) return none as modifies the original array
arr2=arr[::-1]
arr3=np.argsort(arr)# Returns the indices that would sort an array. It does not sort the array itself but gives the indices that would sort the array.
#example: arr=[3,1,9] then np.argsort(arr) will return [1,0,2] as the index of 1 is 1, index of 3 is 0 and index of 9 is 2
a4=np.argmin(arr)
a5=np.argmax(arr)
print(f"Sort the array in ascending order:{arr1} \nSort in descending order:{arr2} \nReturn the indices that would sort the array (np.argsort):{arr3} \nFind the index of the minimum:{a4} and maximum values:{a5} ")
````
