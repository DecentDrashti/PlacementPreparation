## 🗂️ Dataset Setup — Run This First!

The dataset represents **retail sales transactions** for a company.  
It contains **200 rows** and the following columns:

| Column | Description |
|--------|-------------|
| `order_id` | Unique order identifier |
| `order_date` | Date of the order (YYYY-MM-DD) |
| `customer_name` | Full name of the customer |
| `city` | City of the customer |
| `region` | Sales region (North / South / East / West) |
| `category` | Product category (Electronics / Clothing / Food / Furniture) |
| `product` | Product name |
| `quantity` | Number of units ordered |
| `unit_price` | Price per unit (USD) |
| `discount` | Discount percentage applied (0–30%) |
| `salesperson` | Name of the salesperson responsible |
| `rating` | Customer satisfaction rating (1–5) |


````python
import pandas as pd
import numpy as np

np.random.seed(42)
n = 200

cities    = ['New York', 'Los Angeles', 'Chicago', 'Houston', 'Phoenix',
             'Philadelphia', 'San Antonio', 'Dallas', 'Austin', 'Seattle']
regions   = {'New York': 'East', 'Los Angeles': 'West', 'Chicago': 'North',
             'Houston': 'South', 'Phoenix': 'West', 'Philadelphia': 'East',
             'San Antonio': 'South', 'Dallas': 'South', 'Austin': 'South', 'Seattle': 'West'}
categories = ['Electronics', 'Clothing', 'Food', 'Furniture']
products   = {
    'Electronics': ['Laptop', 'Smartphone', 'Headphones', 'Tablet', 'Camera'],
    'Clothing':    ['T-Shirt', 'Jeans', 'Jacket', 'Sneakers', 'Dress'],
    'Food':        ['Coffee Beans', 'Olive Oil', 'Pasta', 'Chocolate', 'Tea'],
    'Furniture':   ['Office Chair', 'Desk', 'Bookshelf', 'Sofa', 'Lamp']
}
prices     = {
    'Laptop': 999.99, 'Smartphone': 799.99, 'Headphones': 199.99,
    'Tablet': 499.99, 'Camera': 649.99, 'T-Shirt': 29.99,
    'Jeans': 59.99, 'Jacket': 89.99, 'Sneakers': 79.99, 'Dress': 69.99,
    'Coffee Beans': 14.99, 'Olive Oil': 12.99, 'Pasta': 5.99,
    'Chocolate': 8.99, 'Tea': 9.99, 'Office Chair': 249.99,
    'Desk': 349.99, 'Bookshelf': 149.99, 'Sofa': 799.99, 'Lamp': 49.99
}
salespersons = ['Alice Johnson', 'Bob Martinez', 'Carol White',
                'David Lee', 'Emma Davis']
first_names  = ['James', 'Mary', 'John', 'Patricia', 'Robert', 'Jennifer',
                'Michael', 'Linda', 'William', 'Barbara', 'David', 'Susan',
                'Richard', 'Jessica', 'Thomas', 'Sarah', 'Charles', 'Karen',
                'Daniel', 'Lisa']
last_names   = ['Smith', 'Johnson', 'Williams', 'Brown', 'Jones', 'Garcia',
                'Miller', 'Davis', 'Wilson', 'Moore', 'Taylor', 'Anderson',
                'Thomas', 'Jackson', 'White', 'Harris', 'Martin', 'Thompson',
                'Young', 'Robinson']

city_list    = np.random.choice(cities, n)
cat_list     = np.random.choice(categories, n)
prod_list    = [np.random.choice(products[c]) for c in cat_list]
price_list   = np.array([prices[p] for p in prod_list])
disc_list    = np.random.choice([0, 5, 10, 15, 20, 25, 30], n)
qty_list     = np.random.randint(1, 11, n)
rating_list  = np.random.choice([1, 2, 3, 4, 5], n, p=[0.05, 0.10, 0.20, 0.40, 0.25])
dates        = pd.date_range(start='2023-01-01', end='2023-12-31', periods=n)
dates        = pd.to_datetime(np.sort(np.random.choice(dates, n, replace=False)))
cust_names   = [f"{np.random.choice(first_names)} {np.random.choice(last_names)}" for _ in range(n)]

# Introduce some missing values
rating_arr = rating_list.astype(float)
rating_arr[np.random.choice(n, 12, replace=False)] = np.nan
disc_arr = disc_list.astype(float)
disc_arr[np.random.choice(n, 8, replace=False)] = np.nan

df = pd.DataFrame({
    'order_id':      [f'ORD-{1000+i}' for i in range(n)],
    'order_date':    dates,
    'customer_name': cust_names,
    'city':          city_list,
    'region':        [regions[c] for c in city_list],
    'category':      cat_list,
    'product':       prod_list,
    'quantity':      qty_list,
    'unit_price':    price_list,
    'discount':      disc_arr,
    'salesperson':   np.random.choice(salespersons, n),
    'rating':        rating_arr
})

# Add duplicate rows intentionally
df = pd.concat([df, df.iloc[[5, 23, 77]]], ignore_index=True)

print(f'✅ Dataset loaded! Shape: {df.shape}')
df.head()
````

### ✅ Task 18 — Apply and Map
1. Use `.apply()` to create a new column `price_category`:
   - `'Budget'` if unit_price < 50
   - `'Mid-range'` if 50 ≤ unit_price < 300
   - `'Premium'` if unit_price ≥ 300
2. Use `.map()` on the `region` column to create a `region_code` column:
   - North → 'N', South → 'S', East → 'E', West → 'W'
3. Print the value counts for both new columns.

````python
# ✏️ Write your answer here
def categorize_price(price):
    if price < 50:
        return 'Budget'
    elif 50 <= price < 300:
        return 'Mid-range'
    else:
        return 'Premium'

df['price_category'] = df['unit_price'].apply(categorize_price)
````

````python
region_mapping = {
    'North': 'N',
    'South': 'S',
    'East': 'E',
    'West': 'W'
}

df['region_code'] = df['region'].map(region_mapping)
````

````python
print("Price Category Counts:")
print(df['price_category'].value_counts())

print("\nRegion Code Counts:")
print(df['region_code'].value_counts())
````

### ✅ Task 20 — Merging DataFrames
Create this separate DataFrame of **salesperson targets**:
```python
targets = pd.DataFrame({
    'salesperson':    ['Alice Johnson', 'Bob Martinez', 'Carol White', 'David Lee', 'Emma Davis'],
    'monthly_target': [15000, 12000, 18000, 10000, 14000],
    'experience_yrs': [5, 3, 8, 2, 6]
})
```
1. **Merge** `df` with `targets` on `salesperson` using an **inner join**.
2. Compute each salesperson's **total revenue** and compare it to their `monthly_target`.
3. Create a column `hit_target` (True/False) — True if total revenue ≥ target.

````python
# ✏️ Write your answer here
targets = pd.DataFrame({
    'salesperson':    ['Alice Johnson', 'Bob Martinez', 'Carol White', 'David Lee', 'Emma Davis'],
    'monthly_target': [15000, 12000, 18000, 10000, 14000],
    'experience_yrs': [5, 3, 8, 2, 6]
})
df_merged = pd.merge(df, targets, on='salesperson', how='inner')
df_merged
````

````python
# Group by salesperson to get total revenue and keep their target
performance = df_merged.groupby(['salesperson', 'monthly_target'])['total_revenue'].sum().reset_index()

# Create the hit_target column
performance['hit_target'] = performance['total_revenue'] >= performance['monthly_target']

# Display results
print(performance)
````
